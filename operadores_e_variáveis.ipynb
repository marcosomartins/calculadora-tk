# Tabela Verdade
## And

```python
print(10 > 2 and 4 < 6)   # V and V - V
print(10 > 2 and 4 > 6)   # V and F - F
print(10 < 2 and 4 < 6)   # F and F - F
print(10 < 2 and 4 > 6)   # F and F - F
```

### Or

```python
print(10 > 2 or 4 < 6)    # V or V - V
print(10 > 2 or 4 > 6)    # V or G - V
print(10 < 2 or 4 < 6)    # F or V - V
print(10 < 2 or 4 > 6)    # F or F - F
```

## Not

```python
not 10 > 2    # V - F
```

```python
not 10 < 2    # F - V
```

# Condicional

```python
# if
if 5 < 10:
  print("verdade")
```

```python
# if/else
num1 = 10
num2 = 5
if num1 > num2:
  print("o num1 é o maior número")

else:
  print("o num2 é o maior número")
```

```python
# if / elief / else
num1 = 10
num2 = 5
if num1 > num2:
  print("o num1 é o maior num")
elif num1 < num2:
  print("o num2 é o maior num")
else:
  print("os dois são iguais")
```

# Condicional com operadores lógicos

```python
num1 = 10
num2 = 5
num3 = 2
if (num1 > num2) and (num1 > num3):
  print("o num1 é maior numero")
else:
  print("o num1 não é maior numero")
```

# Estrutura de repetição

## For

```python
# for range
for contador in range(10):
  print(contador)
```

```python
# for range + break
for contador in range(30):
  print(contador)
  if contador ==2:
    break
```

```python
lista = ["teste1", "teste2"]
```

```python
# for simples
for val in lista:
  print(val)
```

```python
# for com enumerate
for i, val in enumerate(lista):
  print(i, val)
```

## For com operadores compostos

```python
lista2 = [2, 3, 4]
```

```python
for val in lista2:
    val += 1
    print(val)
```

```python
for val in lista2:
    val *= 2
    print(val)
```

## While

```python
val = 1
while (val < 6):
    val*=2
    print(val)
```

```python
# while com else
val = 10
while val < 6:
    val *= 2
else:
    print("valor de val é:", val)
```
